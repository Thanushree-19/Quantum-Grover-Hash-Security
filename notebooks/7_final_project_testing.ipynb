```python

print("=" * 55)
print("CELL 2 - TOY HASH FUNCTION TEST")
print("=" * 55)

# Test configuration
n_bits = 3
m_bits = 2

print(f"\nInput bits: {n_bits}")
print(f"Hash bits: {m_bits}")

# Generate the toy hash system using the existing module
hash_map = toy_hash.create_toy_hash_system(
    n_bits=n_bits,
    m_bits=m_bits
)

# Check the result
assert isinstance(hash_map, dict), (
    "FAIL: Expected the toy hash system to return a dictionary."
)

expected_inputs = 2 ** n_bits

assert len(hash_map) == expected_inputs, (
    f"FAIL: Expected {expected_inputs} inputs, "
    f"but received {len(hash_map)}."
)

print(f"\nPASS: Hash map contains {len(hash_map)} inputs.")

# Display generated mappings
print("\nGenerated hash mappings:")

for input_value, hash_value in hash_map.items():
    print(f"{input_value} -> {hash_value}")

# Check consistency by generating the same configuration again
hash_map_repeat = toy_hash.create_toy_hash_system(
    n_bits=n_bits,
    m_bits=m_bits
)

assert hash_map == hash_map_repeat, (
    "FAIL: Repeated generation produced different mappings."
)

print("\nPASS: Repeated generation is consistent.")
print("CELL 2 COMPLETED")
```

```python

print("=" * 55)
print("CELL 3 - CLASSICAL PREIMAGE SEARCH TEST")
print("=" * 55)

# Use the hash map generated in Cell 2
target_hash = next(iter(hash_map.values()))

print(f"\nTarget hash: {target_hash}")

# Run the existing classical search function
matches = preimage_search.classical_preimage_search(
    hash_map,
    target_hash
)

print("\nMatching inputs:")
print(matches)

# Independently calculate the expected matches
expected_matches = [
    input_value
    for input_value, hash_value in hash_map.items()
    if hash_value == target_hash
]

# Verify the search result
assert set(matches) == set(expected_matches), (
    f"FAIL: Expected {expected_matches}, received {matches}"
)

assert len(matches) > 0, (
    "FAIL: At least one matching input was expected."
)

print(f"\nTarget hash: {target_hash}")
print(f"Expected matches: {expected_matches}")
print(f"Actual matches:   {matches}")
print(f"Number of matches: {len(matches)}")

print("\nPASS: Classical preimage search returned all matches.")
print("CELL 3 COMPLETED")
```

```python

print("=" * 55)
print("CELL 4 - GROVER ORACLE TEST")
print("=" * 55)

# 1. Reuse the hash map and target hash from Cell 3
print(f"\nTarget hash: {target_hash}")
print(f"Expected matching inputs: {expected_matches}")

# 2. Find marked inputs using the existing oracle module
marked_inputs = oracle.find_marked_inputs(
    hash_map,
    target_hash
)

print(f"\nOracle marked inputs: {marked_inputs}")

# 3. Verify the marked inputs against classical search
assert set(marked_inputs) == set(expected_matches), (
    f"FAIL: Expected marked inputs {expected_matches}, "
    f"but received {marked_inputs}"
)

print("PASS: Oracle identifies the correct matching inputs.")

# 4. Build the oracle circuit
# The third argument specifies the number of input bits.
oracle_circuit = oracle.create_grover_oracle(
    hash_map,
    target_hash,
    n_bits
)

# 5. Display the circuit
print("\nOracle circuit:")
print(oracle_circuit.draw(output="text"))

# 6. Verify the number of qubits
assert oracle_circuit.num_qubits == n_bits, (
    f"FAIL: Expected {n_bits} qubits, "
    f"but received {oracle_circuit.num_qubits}"
)

print(f"\nPASS: Oracle circuit uses {n_bits} qubits.")

# 7. Final result
print("\nCELL 4 COMPLETED")
```

```python

import numpy as np

print("=" * 55)
print("CELL 5 - DIFFUSION OPERATOR TEST")
print("=" * 55)

# 1. Create the diffusion operator
diffusion_circuit = grover_search.create_diffusion_operator(
    n_bits
)

print(f"\nNumber of qubits: {diffusion_circuit.num_qubits}")
print("\nDiffusion circuit:")
print(diffusion_circuit.draw(output="text"))

# 2. Verify the number of qubits
assert diffusion_circuit.num_qubits == n_bits, (
    f"FAIL: Expected {n_bits} qubits, "
    f"but received {diffusion_circuit.num_qubits}"
)

print("\nPASS: Correct number of qubits.")

# 3. Convert the circuit to a unitary matrix
from qiskit.quantum_info import Operator

diffusion_operator = Operator(diffusion_circuit)
unitary_matrix = diffusion_operator.data

# 4. Verify unitarity: U†U = I
identity_matrix = np.eye(2 ** n_bits)

assert np.allclose(
    unitary_matrix.conj().T @ unitary_matrix,
    identity_matrix
), "FAIL: Diffusion operator is not unitary."

print("PASS: Diffusion operator is unitary.")

# 5. Verify matrix dimensions
expected_dimension = 2 ** n_bits

assert unitary_matrix.shape == (
    expected_dimension,
    expected_dimension
), "FAIL: Unexpected diffusion matrix dimensions."

print(
    f"PASS: Matrix dimensions are "
    f"{unitary_matrix.shape}."
)

print("\nCELL 5 COMPLETED")
```

```python

import inspect
import numpy as np

print("=" * 55)
print("CELL 6 - GROVER CIRCUIT TEST")
print("=" * 55)

# Inspect the existing function signature before calling it
print("\nGrover-related functions available:")

for name in dir(grover_search):
    if "grover" in name.lower() or "probab" in name.lower():
        obj = getattr(grover_search, name)

        if callable(obj):
            print(f"{name}{inspect.signature(obj)}")

# Find the existing Grover circuit builder
builder_names = [
    "create_grover_circuit",
    "build_grover_circuit",
    "grover_circuit",
]

builder = None

for name in builder_names:
    candidate = getattr(grover_search, name, None)

    if callable(candidate):
        builder = candidate
        print(f"\nCircuit builder found: {name}")
        break

if builder is None:
    print(
        "\nCould not identify the circuit builder automatically."
    )
    print(
        "Share the function names printed above so we can "
        "use your existing implementation correctly."
    )
else:
    print("\nBuilder signature:")
    print(inspect.signature(builder))

    print(
        "\nWe will check the existing builder's parameters "
        "before running the circuit."
    )

print("\nCELL 6 INSPECTION COMPLETED")
```

```python

import numpy as np
from qiskit.quantum_info import Statevector

print("=" * 55)
print("CELL 7 - GROVER CIRCUIT AND PROBABILITY TEST")
print("=" * 55)

# 1. Display test configuration
number_of_matches = len(expected_matches)

print(f"\nInput bits: {n_bits}")
print(f"Total possible inputs: {2 ** n_bits}")
print(f"Target hash: {target_hash}")
print(f"Number of matching inputs: {number_of_matches}")

# 2. Build the existing Grover circuit
grover_circuit = grover_search.create_grover_circuit(
    hash_map,
    target_hash,
    n_bits
)

print("\nPASS: Grover circuit created successfully.")

# 3. Verify qubit count
assert grover_circuit.num_qubits == n_bits, (
    f"Expected {n_bits} qubits, "
    f"got {grover_circuit.num_qubits}"
)

print(f"PASS: Circuit contains {n_bits} qubits.")

# 4. Display the complete circuit
print("\nGrover circuit:")
print(grover_circuit.draw(output="text"))

# 5. Remove measurements from a COPY for statevector simulation
# The original circuit remains unchanged.
simulation_circuit = grover_circuit.remove_final_measurements(
    inplace=False
)

# Verify there are no remaining classical bits
assert simulation_circuit.num_clbits == 0, (
    "FAIL: Circuit still contains classical bits."
)

print("\nPASS: Measurements removed from simulation copy.")

# 6. Calculate statevector probabilities
statevector = Statevector.from_instruction(simulation_circuit)
probabilities = statevector.probabilities()

# 7. Verify probabilities
assert np.isclose(np.sum(probabilities), 1.0), (
    "FAIL: Probabilities do not sum to 1."
)

assert len(probabilities) == 2 ** n_bits, (
    "FAIL: Unexpected number of probabilities."
)

print("PASS: Statevector simulation completed.")
print("PASS: Probabilities sum to 1.")

# 8. Display probability for every input
print("\nStatevector probabilities:")

for input_value, probability in enumerate(probabilities):
    print(f"Input {input_value}: {probability:.6f}")

# 9. Calculate total marked and unmarked probabilities
marked_probability = sum(
    probabilities[input_value]
    for input_value in expected_matches
)

unmarked_probability = sum(
    probability
    for input_value, probability in enumerate(probabilities)
    if input_value not in expected_matches
)

print(f"\nTotal marked-state probability: {marked_probability:.6f}")
print(f"Total unmarked-state probability: {unmarked_probability:.6f}")

assert np.isclose(
    marked_probability + unmarked_probability,
    1.0
), "FAIL: Probability totals are inconsistent."

print("\nPASS: Marked and unmarked probabilities are valid.")
print("CELL 7 COMPLETED")
```

```python

import numpy as np
from qiskit.quantum_info import Statevector

print("=" * 55)
print("CELL 8 - END-TO-END INTEGRATION TEST")
print("=" * 55)

# 1. Generate a fresh toy hash system
test_n_bits = 3
test_m_bits = 2

integration_hash_map = toy_hash.create_toy_hash_system(
    n_bits=test_n_bits,
    m_bits=test_m_bits
)

assert len(integration_hash_map) == 2 ** test_n_bits
print("PASS: Toy hash generation.")

# 2. Select a target hash
integration_target = next(iter(integration_hash_map.values()))

# 3. Run classical preimage search
integration_matches = preimage_search.classical_preimage_search(
    integration_hash_map,
    integration_target
)

assert len(integration_matches) > 0
print(f"PASS: Classical search found {len(integration_matches)} matches.")

# 4. Verify oracle marking
integration_marked = oracle.find_marked_inputs(
    integration_hash_map,
    integration_target
)

assert set(integration_marked) == set(integration_matches)
print("PASS: Oracle matches classical search.")

# 5. Build the Grover circuit
integration_circuit = grover_search.create_grover_circuit(
    integration_hash_map,
    integration_target,
    test_n_bits
)

assert integration_circuit.num_qubits == test_n_bits
print("PASS: Grover circuit construction.")

# 6. Simulate without final measurements
integration_simulation = integration_circuit.remove_final_measurements(
    inplace=False
)

integration_state = Statevector.from_instruction(
    integration_simulation
)

integration_probabilities = integration_state.probabilities()

# 7. Validate probability distribution
assert np.isclose(integration_probabilities.sum(), 1.0)
assert len(integration_probabilities) == 2 ** test_n_bits

print("PASS: Statevector simulation.")
print("PASS: Probability normalization.")

# 8. Calculate total marked-state probability
integration_success_probability = sum(
    integration_probabilities[index]
    for index in integration_matches
)

print(f"\nTarget hash: {integration_target}")
print(f"Classical matches: {integration_matches}")
print(f"Oracle matches: {integration_marked}")
print(
    "Grover marked-state probability: "
    f"{integration_success_probability:.6f}"
)

print("\nALL INTEGRATION CHECKS PASSED.")
print("CELL 8 COMPLETED")
```

```python

print("=" * 55)
print("CELL 9 - EDGE CASE TESTING")
print("=" * 55)

# Test 1: Target with multiple matches
print("\nTEST 1: Multiple matching inputs")

multiple_target = target_hash

multiple_matches = preimage_search.classical_preimage_search(
    hash_map,
    multiple_target
)

multiple_marked = oracle.find_marked_inputs(
    hash_map,
    multiple_target
)

assert len(multiple_matches) >= 2, (
    "Expected at least two matches for this test."
)

assert set(multiple_matches) == set(multiple_marked), (
    "Classical search and oracle disagree."
)

print(f"PASS: Multiple matches found: {multiple_matches}")

# Test 2: Find a target with exactly one matching input
print("\nTEST 2: Exactly one matching input")

unique_targets = [
    value
    for value in hash_map.values()
    if list(hash_map.values()).count(value) == 1
]

if unique_targets:
    unique_target = unique_targets[0]

    unique_matches = preimage_search.classical_preimage_search(
        hash_map,
        unique_target
    )

    unique_marked = oracle.find_marked_inputs(
        hash_map,
        unique_target
    )

    assert len(unique_matches) == 1
    assert set(unique_matches) == set(unique_marked)

    print(f"PASS: Unique target {unique_target}")
    print(f"Matching input: {unique_matches}")
else:
    print(
        "SKIP: This hash map has no target with exactly one match."
    )

# Test 3: Target absent from the hash map
print("\nTEST 3: Nonexistent target hash")

possible_hash_values = set(hash_map.values())
all_hash_values = set(range(2 ** m_bits))

missing_targets = all_hash_values - possible_hash_values

if missing_targets:
    missing_target = next(iter(missing_targets))

    missing_matches = preimage_search.classical_preimage_search(
        hash_map,
        missing_target
    )

    missing_marked = oracle.find_marked_inputs(
        hash_map,
        missing_target
    )

    assert len(missing_matches) == 0
    assert len(missing_marked) == 0

    print(f"PASS: Target {missing_target} has no matching inputs.")
else:
    print(
        "SKIP: Every possible hash value appears in the hash map."
    )

print("\nCELL 9 EDGE-CASE TESTS COMPLETED")
```
