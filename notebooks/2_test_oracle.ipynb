```python

import sys
from pathlib import Path

# Locate the project root.
project_root = Path.cwd().parent

# Make sure Python can find the project.
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

# Import the existing project modules.
from src.hash_functions.toy_hash import create_toy_hash_system
from src.quantum_grover.oracle import (
    find_marked_inputs,
    create_grover_oracle,
)

print("All imports successful!")
print("Project root:", project_root)
```

```python

n_bits = int(input("Enter input bit size (2 to 8): "))
m_bits = int(input("Enter hash output size (in bits): "))

if not 2 <= n_bits <= 8:
    raise ValueError("Input bit size must be between 2 and 8.")

if m_bits < 1:
    raise ValueError("Hash output size must be at least 1.")

print(f"Input bits: {n_bits}")
print(f"Hash bits: {m_bits}")
print(f"Possible inputs: {2**n_bits}")
```

```python

hash_map = create_toy_hash_system(n_bits, m_bits)

print("\nHash map created.")
print("Total inputs:", len(hash_map))
```

```python

target_hash = int(input("Enter target hash (decimal): "))

if not 0 <= target_hash < 2**m_bits:
    raise ValueError("Target hash is outside the selected hash size.")

marked_inputs = find_marked_inputs(hash_map, target_hash)

print("\nTarget hash:", target_hash)
print("Matching inputs:", marked_inputs)
print("Number of matching inputs:", len(marked_inputs))
```
